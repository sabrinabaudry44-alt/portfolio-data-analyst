3.
Les statistiques descriptives permettent d'obtenir une première vision du dataset nettoyé. Elles résument les principales caractéristiques des variables numériques et facilitent l'identification des tendances générales, des dispersions et des valeurs extrêmes.

Cette étape a pour objectif de :

vérifier la cohérence des données après le nettoyage ;
observer la répartition des principales variables (population, PIB, espérance de vie, taux de chômage, etc.) ;
comparer les ordres de grandeur des indicateurs ;
identifier les éventuelles valeurs atypiques qui pourraient influencer l'analyse.
Pour réaliser cette analyse, nous utilisons notamment les fonctions describe(), mean(), median(), min(), max() et std(), qui fournissent des indicateurs tels que :

la moyenne : valeur moyenne de la variable ;
la médiane : valeur centrale, moins influencée par les valeurs extrêmes ;
l'écart-type : mesure de la dispersion des données autour de la moyenne ;
les valeurs minimale et maximale : permettent d'identifier les amplitudes des variables ;
les quartiles : donnent une vision de la répartition des données.
L'analyse de ces indicateurs constitue une première étape indispensable avant les analyses de corrélation et les visualisations, car elle permet de mieux comprendre la structure du jeu de données et d'interpréter les résultats obtenus par la suite.

3.2 Distribution des principales variables

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Variables numériques à représenter
variables = [
    "GDP",
    "Population",
    "Life expectancy",
    "Unemployment rate",
    "Birth rate",
    "Fertility rate",
    "Infant mortality",
    "Maternal mortality ratio",
    "CPI Change (%)"
]

# Histogrammes
for variable in variables:

    # Vérifie que la colonne existe
    if variable not in df.columns:
        print(f"Colonne introuvable : {variable}")
        continue

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df,
        x=variable,
        bins=30,
        kde=True
    )

    plt.title(f"Distribution de {variable}")
    plt.xlabel(variable)
    plt.ylabel("Nombre de pays")

    plt.tight_layout()
    plt.show()
```

3.5 Répondre à la problématique métier

Exemple :

Les analyses montrent que les pays présentant un PIB élevé possèdent généralement une espérance de vie plus importante ainsi qu'un taux d'accès à Internet supérieur. À l'inverse, les pays présentant un taux de chômage élevé affichent souvent des indicateurs économiques moins favorables. Ces observations suggèrent que plusieurs variables économiques et sociales peuvent être utilisées pour caractériser le potentiel d'attractivité d'un pays.

```python
from pathlib import Path

path = Path("world-data-2023-clean.csv")

if not path.exists():
    path = Path("../data/world-data-2023-clean.csv")

df = pd.read_csv(path)
```

```python
import math
import matplotlib.pyplot as plt
import seaborn as sns

variables_box = df.select_dtypes(include="number").columns

n = len(variables_box)
ncols = 2
nrows = math.ceil(n / ncols)

plt.figure(figsize=(12, 4 * nrows))

for i, variable in enumerate(variables_box, 1):
    plt.subplot(nrows, ncols, i)
    sns.boxplot(y=df[variable].dropna())
    plt.title(variable)

plt.tight_layout()
plt.show()
```

Les boxplots montrent la présence de plusieurs valeurs extrêmes, notamment pour le PIB et la population. Ces observations correspondent à des pays ayant une économie ou une population exceptionnellement importantes. Des valeurs extrêmes sont également observées pour le taux de chômage, l'inflation et les indicateurs de santé. Ces valeurs ne sont pas considérées comme des erreurs de saisie mais reflètent des différences réelles entre les pays. Elles ont donc été conservées dans l'analyse.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="GDP",
    y="Life expectancy"
)

plt.title("Relation entre le PIB et l'espérance de vie")
plt.xlabel("PIB")
plt.ylabel("Espérance de vie")

plt.show()
```

Ce graphique permet d'observer si les pays ayant un PIB élevé bénéficient également d'une espérance de vie plus importante. Une tendance positive suggérerait qu'un niveau de richesse élevé est souvent associé à de meilleures conditions de vie.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="GDP",
    y="Unemployment rate"
)

plt.title("Relation entre le PIB et le taux de chômage")
plt.xlabel("PIB")
plt.ylabel("Taux de chômage (%)")

plt.show()
```
